```python
new_sales = [5000, 2500, 1700]
```

```python
with open("sales_data.txt", "a") as file:
    for amount in new_sales:
        file.write(str(amount) + "\n")
```

```python
with open("sales_data.txt", "r") as file:
    print(file.read())
```

```python
with open("sales_data.txt", "r") as file:
    lines = file.readlines()
    print("Total number of lines:", len(lines))
```
