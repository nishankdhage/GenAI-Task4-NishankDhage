```python
prices = {
    "Mouse": 500,
    "Keyboard": 800,
    "Monitor": 7000,
    "Pendrive": 400,
    "Camera": 5000
}
```

```python
discount = float(input("Enter discount percentage: "))
```

```python
discounted_prices = []
```

```python
for product, price in prices.items():
    discounted_price = price - (price * discount / 100)

    discounted_prices.append(discounted_price)
```

```python
with open("discount_report.txt", "w") as file:

    file.write("Product | Original Price | Discounted Price\n")

    for product, price in prices.items():
        discounted_price = price - (price * discount / 100)

        file.write(
            product + " | " +
            str(price) + " | " +
            str(round(discounted_price, 2)) + "\n"
        )
```

```python
with open("discount_report.txt", "r") as file:
    print(file.read())
```

```python
total_items = len(discounted_prices)

average_price = sum(discounted_prices) / total_items

print("Total Items:", total_items)
print("Average Discounted Price:", round(average_price, 2))
```
